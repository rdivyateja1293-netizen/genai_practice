```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])

print("Vector A:", vector_a)
print("Vector B:", vector_b)
```

```python
dot_product = np.dot(vector_a, vector_b)

magnitude_a = np.linalg.norm(vector_a)
magnitude_b = np.linalg.norm(vector_b)

cosine_similarity_value = dot_product / (magnitude_a * magnitude_b)

print("Dot Product:", dot_product)
print("Magnitude of A:", magnitude_a)
print("Magnitude of B:", magnitude_b)
print("Cosine Similarity:", cosine_similarity_value)
```

```python
result = cosine_similarity(
    vector_a.reshape(1, -1),
    vector_b.reshape(1, -1)
)

print("Cosine Similarity using Scikit-learn:", result[0][0])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

sentences = [
    "I love machine learning",
    "I love artificial intelligence"
]

vectorizer = CountVectorizer()
vectors = vectorizer.fit_transform(sentences)

similarity = cosine_similarity(vectors)

print("Cosine Similarity Matrix:")
print(similarity)
```

```python
import matplotlib.pyplot as plt

plt.bar(
    ["Mathematical", "Scikit-learn"],
    [cosine_similarity_value, result[0][0]]
)

plt.ylabel("Cosine Similarity")
plt.title("Cosine Similarity Comparison")
plt.ylim(0, 1.1)
plt.show()
```
